# **Análise de Variáveis Socioeconômicas de Bangladesh**

- **Pacotes python**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


pd.set_option('display.max_rows', None)

pd.set_option('display.max_columns', None)
```

- **Carregando os dados**

```python

URL = 'https://raw.githubusercontent.com/Samuelsilvestre/Bangladesh/master/dados/Literacy%20Rate%20inflation%20GDP%20in%20Bangladesh%20Dataset%20-%20RAW.csv'

data = pd.read_csv(URL, sep=',', encoding='utf-8')

```

- **Estrutura dos dados**

```python
print(f' Linhas {data.shape[0]}')
print(f' Colunas {data.shape[1]}')
```

- **Nome das colunas**

```python
data.columns
```

 - **Valores nulos por colunas**

```python
null = data.isnull().sum()
sns.heatmap(data.isnull(), cbar=False)
print(null)
plt.show()
```

- **Tipo de dados**

```python
data.dtypes
```

- **Selecionando as colunas mais relevantes**

```python
df = data[[
    'Year',
    'Literacy Rate(%)',
    'Population ',
    'Population growth rate (%)',
    'Migrant Population (Estimated)',
    'Net Migration Rate',
    'Growth Rate',
    'Inflation Rate (%)',
    'Annual Change (%)',
    'GNI (Billion USD)'
]].copy()

```

- **Renomeando as colunas**

```python
new_name = [
    'Ano',
    'Taxa de Alfabetização (%)',
    'População',
    'Taxa de Crescimento Populacional (%)',
    'População Migrante (Estimada)',
    'Taxa de Migração Líquida',
    'Taxa de Crescimento (Geral)',
    'Taxa de Inflação (%)',
    'Mudança Anual (%) (Inflação)',
    'RNB (Bilhões de USD)'
]

df.columns = new_name
```

- **Estrutura atual dos dados**

```python
display(df.head(10))
display(df.dtypes)
display(df.isnull().sum())
```

- **Deletando as linhas nulas**

```python
df.dropna(how='all', axis=0, inplace=True)
print(f'Linhas {df.shape[0]}')
print(f'Colunas {df.shape[1]}')
```

```python
display(df.head(10))
display(df.dtypes)
display(df.isnull().sum())
```

- **Tratamento das variáveis numericas**

```python
for col in df.columns:
  if df[col].astype(str).str.contains('%').any():
    print(f'{col} tipo {df[col].dtype}')

    df[col] = df[col].apply(
        lambda x: str(x).split('%')[0] if pd.notnull(x) else None
    )
    df[col] = df[col].astype('float64')

    print(f'{col} apos tratamento {df[col].dtype}')
```

- **Tratando as colunas com o tipo object**

```python

for col in df.columns:
    if df[col].dtype == object:
        print(f'Processando coluna object: {col}')

        df[col] = df[col].apply(
            lambda x: str(x).replace(',','').replace('.','')
        )
        df[col] = df[col].astype('int64')

```

- **Convertendo o tipo da coluna Ano e tornando em indice**

```python
df['Ano'] = df['Ano'].astype('int64')
df.set_index('Ano', inplace=True)
```

- **Deletendo linhas nulas**

```python
df.dropna(inplace=True)
```

 - **Nova estrutura dos dados**

```python
display(df.shape)
display(df.head(10))
display(df.dtypes)
display(df.isnull().sum())
```

# Visualização dos indicadores sociais

- **Série Histórica de Bangladesh**

```python

sns.set_theme(style="darkgrid")

plt.figure(figsize=(18, 12))

for col in df.columns:
    if df[col].dtype in ['float64', 'int64']:

        sns.lineplot(data=df[col], color='#000080', linewidth=4)

        plt.title(f'Série {col} em Bangladesh')
        plt.xlabel('Ano')
        plt.ylabel(f'{col}')
        plt.savefig(f'{col}_serie_historica.png', dpi=400)
        plt.savefig(f'./../fig/serie_{col}.png', dpi=400)
        plt.show()
```

- **Matriz de Dispersão das Variáveis de Bangladesh**

```python
sns.set_style('darkgrid')
sns.pairplot(df)
plt.title('Matriz de Dispersão das Variáveis de Bangladesh')
plt.savefig('./../fig/dispersao.png', dpi=400)
plt.show();
```

- **Matriz de Correlação das Variáveis de Bangladesh**

```python
plt.figure(figsize=(12, 10))

corr = df.corr()

sns.heatmap(corr,
            annot=True,
            fmt=".2f",
            cmap='coolwarm',
            cbar=True,
            linewidths=.5,
            linecolor='black')

plt.suptitle('Matriz de Correlação das Variáveis de Bangladesh', y=1.02, fontsize=16)
display(corr)
plt.savefig('./../fig/correlacao.png', dpi=400)
plt.show()
```

- **Classificando as decadas**

```python
bins = [1979, 1989, 1999, 2009, 2019, 2023]

labels = [
    'Década de 80',
    'Década de 90',
    'Década de 00',
    'Década de 10',
    'Década de 20'
]

df['Decada'] = pd.cut(
    x=df.index,
    bins=bins,
    labels=labels,
    right=True,
    include_lowest=False,
    ordered=True
)
```

- **Crescimento Econômico Médio por Década**

```python
dec_mean = df.groupby(['Decada'])['RNB (Bilhões de USD)'].mean()
index = dec_mean.index
values = dec_mean.values

plt.figure(figsize=(20,12))
plt.bar(index, values, color='green')
plt.plot(index, values, color='blue')
plt.title('Crescimento Médio RNB (Bilhões de USD) por Decada', fontsize=18)
plt.tight_layout();
plt.savefig('./../fig/rmb_medio_decada.png', dpi=400)
plt.show();
```

### Conclusão ###
> **O gráfico ilustra que a virada do milênio (a partir de 2000) marcou uma inflexão na trajetória econômica de Bangladesh, saindo de um crescimento lento nas décadas de 80 e 90 para uma fase de crescimento acelerado e exponencial.**
